```python
df_copy_scale=df_copy.copy()
```

```python
df_copy['Age']=2026-df_copy['Yr_built']
df_copy=df_copy.drop(columns='Yr_built')
df_copy_scale['Age']=2026-df_copy_scale['Yr_built']
df_copy_scale=df_copy_scale.drop(columns='Yr_built')
x=df_copy.drop(columns='Price')
y=df_copy['Price']
```

```python
plt.figure(figsize=(14,9))
sns.heatmap(df_copy.corr(numeric_only=True), annot=True)
```

Feature Scaling used for linear regression

```python
from sklearn.preprocessing import StandardScaler
feature_cols=['Bedrooms', 'Bathrooms', 'Sqft_living', 'Floors', 'Waterfront', 'View', 'Age', 'Is_renovated']
scaler=StandardScaler()
df_copy_scale[feature_cols]=scaler.fit_transform(df_copy_scale[feature_cols])
```

```python
df_copy_scale
```

```python
x_scale=df_copy_scale.drop(columns='Price')
y_scale=df_copy_scale['Price']
```

```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score,mean_squared_error,mean_absolute_error
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
x_train_scaled, x_test_scaled, y_train_scaled, y_test_scaled = train_test_split(x_scale, y_scale, test_size=0.2, random_state=42)
```

```python
label=['LinearRegression', 'DecisionTreeRegressor', 'RandomForestRegressor']
values=[]
```

Linear Regression Model

```python
linear_reg_model=LinearRegression()
linear_reg_model.fit(x_train_scaled,y_train_scaled)
```

```python
y_predict=linear_reg_model.predict(x_test_scaled)
```

```python
r2=r2_score(y_test_scaled,y_predict)
values.append(r2)
mse=mean_squared_error(y_test_scaled,y_predict)
mae=mean_absolute_error(y_test_scaled,y_predict)
```

```python
print(r2)
print(mse)
print(mae)
```

Decision Tree

```python
tree_model= DecisionTreeRegressor(random_state=42, max_depth=10)
  # Optional: max_depth helps prevent overfitting
tree_model.fit(x_train,y_train)
y_predict=tree_model.predict(x_test)
```

```python
r2=r2_score(y_test,y_predict)
values.append(r2)
mse=mean_squared_error(y_test,y_predict)
mae=mean_absolute_error(y_test,y_predict)
print(r2)
print(mse)
print(mae)
```

Random Forest

```python
forest_model=RandomForestRegressor(n_estimators=200,random_state=42)
forest_model.fit(x_train, y_train)
```

```python
y_predict=forest_model.predict(x_test)
```

```python
r2=r2_score(y_test,y_predict)
values.append(r2)
mse=mean_squared_error(y_test,y_predict)
mae=mean_absolute_error(y_test,y_predict)
print(r2)
print(mse)
print(mae)
```
